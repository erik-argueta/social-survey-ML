# Exploratory Data Analysis

## Subsetting the Dataset
I want to explore the data that has to do with Education and socio-economic columns. The initial subset contains: 
* Year of the Respondent's response
* Years of Education completed
* Highest Degree completed
* Whether the Respondent was born in the United States

```python
educ_trends = survey.filter(items=["ID", "YEAR", "EDUC", "DEGREE", "BORN"])
```

```python
educ_trends.dtypes
```

## Dictionary Mapping
Due to the survey's response selection, I mapped the responses to be easily utilized as I explored and graphed data. I created new columns with the mappings so as to leave the original imported columns untouched.

```python
DEGREE_labels = {
    0: "Less_HS",
    1: "HS",
    2: "JRorAssc",
    3: "Bach",
    4: "Grad",
    8: "Unknown",
    9: "No_Answer",
    # 7: "NOT APPL"
}

BORN_labels = {
    # 0: "Not Appl",
    1: "Yes",
    2: "No",
    8: "Unknown",
    9: "No_Answer",
}

educ_trends["Response_Year"] = pd.Categorical(educ_trends["YEAR"])

educ_trends["Highest_Degree"] = educ_trends["DEGREE"].map(DEGREE_labels)
educ_trends["Highest_Degree"] = pd.Categorical(educ_trends["Highest_Degree"])

educ_trends["US_Born"] = educ_trends["BORN"].map(BORN_labels)
educ_trends["US_Born"] = pd.Categorical(educ_trends["US_Born"])
```

```python
educ_trends.dtypes
```

```python
( 
    ggplot(educ_trends)
        + geom_bar(aes(y = "Response_Year"), orientation="y")
        + labs(
            title="Number of Respondents per Year",
            x="Total of Respondents",
            y="Year",
        )
)
```

```python
(
    ggplot(educ_trends)
    + geom_bar(aes(y="YEAR", fill="Highest_Degree"), orientation="y",)# position="fill")
    + labs (
        title="Types of Degrees per Respondent per Year",
        x="Total Count",
        y="Year",
        fill="Highest Degree"
    )
    + scale_fill_discrete(
        labels={
                "Bach" : "Bachelors",
                "Less_HS" : "Less than High School",
                "HS" : "High School",
                "Grad" : "Graduate",
                "JRorAssc" : "Junior College or Associates Degree",
                "No_Answer" : "No Answer"
        }
    )
)
```
