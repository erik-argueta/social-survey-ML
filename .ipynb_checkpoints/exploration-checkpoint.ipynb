# Exploratory Data Analysis

```python
educ_trends = survey.filter(items=["ID", "YEAR", "EDUC", "DEGREE", "BORN"])
```

```python
DEGREE_labels = {
    0: "LESS-HS",
    1: "HS",
    2: "JR-ASSC",
    3: "BACH",
    4: "GRAD",
    8: "DON'T KNOW",
    9: "NO ANSWER",
    # 7: "NOT APPL"
}

BORN_labels = {
    # 0: "Not Appl",
    1: "Yes",
    2: "No",
    8: "Don't Know",
    9: "No Answer",
}

educ_trends["YEAR"] = pd.Categorical(educ_trends["YEAR"])

educ_trends["BORN_CAT"] = educ_trends["BORN"].map(BORN_labels)
educ_trends["BORN_CAT"] = pd.Categorical(educ_trends["BORN_CAT"])

educ_trends["DEGREE_CAT"] = educ_trends["DEGREE"].map(DEGREE_labels)
educ_trends["DEGREE_CAT"] = pd.Categorical(educ_trends["DEGREE_CAT"])
```

```python
( 
    ggplot(educ_trends)
        + geom_bar(aes(x = "YEAR"))
)
```
